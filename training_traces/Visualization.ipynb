```python
import numpy as np
import matplotlib.pyplot as plt

# ======================
# USER INPUT
# ======================
npy_file = r"D:\spgxslb-hackathon-p2-seismic-denoising-challenge\training_traces\164_iline.npy"  # path to your npy file

# ======================
# LOAD DATA
# ======================
data = np.load(npy_file)  # shape: (n_samples, n_traces)

print("Data shape:", data.shape)
print("Min:", np.min(data), "Max:", np.max(data))

# ======================
# VISUALIZE
# ======================
plt.figure(figsize=(10, 6))
plt.imshow(
    data,                # transpose if traces are columns
    cmap="seismic",        # common colormap for seismic
    aspect="auto",
    interpolation="none",
    vmin=-np.max(np.abs(data)),  # symmetric color scale
    vmax=np.max(np.abs(data))
)
plt.colorbar(label="Amplitude")
plt.title("Seismic Inline Section")
plt.xlabel("Trace number")
plt.ylabel("Time/depth sample")
plt.gca().invert_yaxis()  # seismic convention: time increases downward
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display
import glob
import os

# ======================
# USER INPUT
# ======================
data_dir = r"D:\spgxslb-hackathon-p2-seismic-denoising-challenge\training_traces"   # path to folder with .npy files
cmap_choice = "seismic"                 # colormap for visualization

# ======================
# LOAD ALL .NPY FILES
# ======================
file_list = sorted(glob.glob(os.path.join(data_dir, "*.npy")))

if not file_list:
    raise FileNotFoundError(f"No .npy files found in {data_dir}")

# load all npy files into memory
sections = [np.load(f) for f in file_list]
names = [os.path.basename(f) for f in file_list]

print(f"Loaded {len(sections)} inline sections:")
for n in names:
    print("  ", n)

# ======================
# SLIDER FUNCTION
# ======================
def show_inline(idx):
    plt.figure(figsize=(10, 6))
    data = sections[idx]
    vmax = np.max(np.abs(data))
    plt.imshow(
        data,
        cmap=cmap_choice,
        aspect="auto",
        vmin=-vmax,
        vmax=vmax
    )
    plt.gca().invert_yaxis()
    plt.title(f"Inline: {names[idx]}")
    plt.xlabel("Trace number")
    plt.ylabel("Time/Depth sample")
    plt.show()

# ======================
# INTERACTIVE SLIDER
# ======================
slider = widgets.IntSlider(min=0, max=len(sections)-1, step=1, value=0, description="Inline")
widgets.interact(show_inline, idx=slider)
```
